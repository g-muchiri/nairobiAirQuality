```python
import psycopg2
import requests
import pandas as pd
import os
from dotenv import load_dotenv

load_dotenv()
```

```python
##Extracting Loation data in Nairobi Location using coordinates
from datetime import datetime

base_url = os.getenv('BASE_URL')

location_url = f"{base_url}/locations"

my_key = os.getenv('API_KEY')
header ={
   "X-API-Key": my_key 
}

params = {
    "coordinates":"-1.278560,36.821249",
    "radius":25000,
    "limit":1000
}

location_response = requests.get(url=location_url, headers=header, params=params)


```

```python
location_response.json()
```

```python
location_data = pd.json_normalize(location_response.json()['results'])
location_data = pd.DataFrame(location_data)
location_data
```

```python
## Exploding data is normalizing a column within a normalized dataframe
exploded_location_data = location_data.explode(column='sensors')
exploded_location_data
```

```python
normalized_location_data =pd.json_normalize(exploded_location_data['sensors']).add_prefix('sensor.')
final_location_data = exploded_location_data.drop(columns=['sensors']).join(normalized_location_data)
final_location_data.columns
```

```python
sensor_location_data = final_location_data[['id', 'name','sensor.id', 'sensor.parameter.name', 'sensor.parameter.units','coordinates.latitude', 'coordinates.longitude']]
sensor_location_data
```
